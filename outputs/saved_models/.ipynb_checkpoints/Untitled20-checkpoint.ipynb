```python
import pandas as pd
import numpy as np
import re
import string
import matplotlib.pyplot as plt
import seaborn as sns
import os
import time
import joblib
import json
import itertools
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report, accuracy_score,
    confusion_matrix, ConfusionMatrixDisplay
)
from sklearn.preprocessing import LabelEncoder

import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Embedding, LSTM, Dense, Dropout, Bidirectional,
    Attention, GlobalAveragePooling1D, Flatten, BatchNormalization
)
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.regularizers import l2

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import gensim.downloader as api

# === 1. Load Dataset ===
df = pd.read_csv('dataset_undersampled.csv')
df['content'] = df['content'].astype(str)

# === 2. Preprocessing ===
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def clean_text(text):
    text = text.lower()
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#\w+', '', text)
    text = re.sub(r'http\S+', '', text)
    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'\s{2,}', ' ', text)
    return stemmer.stem(text.strip())

df['clean_text'] = df['content'].apply(clean_text)

# === 3. Tokenization & Padding ===
MAX_NUM_WORDS = 10000
MAX_SEQUENCE_LENGTH = 100

tokenizer = Tokenizer(num_words=MAX_NUM_WORDS, oov_token="<OOV>")
tokenizer.fit_on_texts(df['clean_text'])

sequences = tokenizer.texts_to_sequences(df['clean_text'])
X = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGTH)

# === 4. Label Encoding ===
le = LabelEncoder()
y = le.fit_transform(df['label'])

# === 5. Split Data ===
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# === 6. Load Word2Vec ===
print("Loading Word2Vec model...")
word2vec_model = api.load("word2vec-google-news-300")
EMBEDDING_DIM = 300
word_index = tokenizer.word_index
embedding_matrix = np.zeros((min(MAX_NUM_WORDS, len(word_index) + 1), EMBEDDING_DIM))

for word, i in word_index.items():
    if i >= MAX_NUM_WORDS:
        continue
    try:
        embedding_vector = word2vec_model[word]
        embedding_matrix[i] = embedding_vector
    except KeyError:
        embedding_matrix[i] = np.random.normal(0, np.sqrt(0.25), EMBEDDING_DIM)

# === 7. Hyperparameters ===
EPOCHS = 20
BATCH_SIZE = 64
LSTM_UNITS = 64
DROPOUT_RATE = 0.4
LEARNING_RATE = 0.001
PATIENCE = 3

# === 8. Model Builder ===
def build_model(use_attention=False, use_dropout=False, bidirectional=False, use_word2vec=False):
    inputs = Input(shape=(MAX_SEQUENCE_LENGTH,))
    
    if use_word2vec:
        embedding_layer = Embedding(
            input_dim=embedding_matrix.shape[0],
            output_dim=embedding_matrix.shape[1],
            weights=[embedding_matrix],
            input_length=MAX_SEQUENCE_LENGTH,
            trainable=False
        )(inputs)
    else:
        embedding_layer = Embedding(
            input_dim=MAX_NUM_WORDS,
            output_dim=128,
            input_length=MAX_SEQUENCE_LENGTH
        )(inputs)

    x = embedding_layer
    x = BatchNormalization()(x)

    lstm_kwargs = {
        "return_sequences": use_attention,
        "dropout": DROPOUT_RATE,
        "recurrent_dropout": DROPOUT_RATE,
        "kernel_regularizer": l2(0.001),
        "recurrent_regularizer": l2(0.001),
    }

    if bidirectional:
        x = Bidirectional(LSTM(LSTM_UNITS, **lstm_kwargs))(x)
    else:
        x = LSTM(LSTM_UNITS, **lstm_kwargs)(x)

    if use_attention:
        attention = Attention()([x, x])
        x = GlobalAveragePooling1D()(attention)
    else:
        x = Flatten()(x)

    if use_dropout:
        x = Dropout(DROPOUT_RATE)(x)

    outputs = Dense(1, activation='sigmoid', kernel_regularizer=l2(0.001))(x)

    model = Model(inputs, outputs)
    optimizer = Adam(learning_rate=LEARNING_RATE)
    model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy'])
    return model

# === 9. Train-Evaluate Function ===
def train_and_evaluate(use_attention=False, use_dropout=False, bidirectional=False, use_word2vec=False, scenario_num=0):
    model = build_model(use_attention, use_dropout, bidirectional, use_word2vec)
    early_stop = EarlyStopping(monitor='val_loss', patience=PATIENCE, restore_best_weights=True)

    # Timing: training
    start_train = time.time()
    history = model.fit(
        X_train, y_train,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        validation_split=0.25,
        verbose=0
    )
    end_train = time.time()
    train_time = end_train - start_train

    # Timing: testing
    start_test = time.time()
    y_pred_prob = model.predict(X_test)
    y_pred = (y_pred_prob > 0.5).astype("int32")
    end_test = time.time()
    test_time = end_test - start_test

    # Akurasi dan laporan
    acc = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred, target_names=le.classes_, output_dict=True)
    
    # Simpan model
    os.makedirs('saved_models', exist_ok=True)
    model.save(f'saved_models/model_scenario_{scenario_num}.h5')

    # Simpan history
    os.makedirs('training_histories', exist_ok=True)
    pd.DataFrame(history.history).to_csv(f'training_histories/scenario_{scenario_num}_history.csv', index=False)

    # Simpan confusion matrix
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=le.classes_)
    disp.plot(cmap='Blues', values_format='d')
    plt.title(f'Confusion Matrix - Scenario {scenario_num}')
    plt.savefig(f'training_histories/scenario_{scenario_num}_confusion_matrix.png')
    plt.close()

    return acc, history, train_time, test_time, report, scenario_num

# === 10. Definisikan Skenario ===
feature_options = list(itertools.product([False, True], repeat=4))
skenario_list = [
    {
        "attention": combo[0],
        "dropout": combo[1],
        "bidirectional": combo[2],
        "word2vec": combo[3]
    }
    for combo in feature_options
]

# === 11. Jalankan Semua Skenario ===
results = []
histories = []
reports = []

for i, s in enumerate(skenario_list, 1):
    print(f"▶ Skenario {i}: {s}")
    acc, history, train_time, test_time, report, sk_num = train_and_evaluate(
        use_attention=s["attention"],
        use_dropout=s["dropout"],
        bidirectional=s["bidirectional"],
        use_word2vec=s["word2vec"],
        scenario_num=i
    )
    results.append({
        "scenario": i,
        "accuracy": acc,
        "train_time_sec": train_time,
        "test_time_sec": test_time,
        **s
    })
    histories.append((i, history))
    reports.append((i, report))

# === 12. Simpan Semua Hasil ===
results_df = pd.DataFrame(results)
results_df.to_csv("all_results.csv", index=False)

# === 13. Temukan Model Terbaik ===
best_result = results_df.sort_values(by="accuracy", ascending=False).iloc[0]
best_idx = best_result["scenario"]
best_report = dict(reports[best_idx - 1][1])

# Simpan classification report terbaik
with open(f"best_model_classification_report_scenario_{best_idx}.json", "w") as f:
    json.dump(best_report, f, indent=4)

print("\n✅ Model terbaik:")
print(f"Skenario {best_idx} dengan akurasi {best_result['accuracy']:.4f}")
```

```python
# === 14. Plot Training History ===
def plot_training_history(histories):
    for i, history in histories:
        plt.figure(figsize=(10, 4))
        plt.subplot(1, 2, 1)
        plt.plot(history.history['accuracy'], label='train acc')
        plt.plot(history.history['val_accuracy'], label='val acc')
        plt.title(f'Skenario {i} - Akurasi')
        plt.xlabel('Epoch')
        plt.ylabel('Accuracy')
        plt.legend()
        plt.subplot(1, 2, 2)
        plt.plot(history.history['loss'], label='train loss')
        plt.plot(history.history['val_loss'], label='val loss')
        plt.title(f'Skenario {i} - Loss')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        plt.tight_layout()
        plt.savefig(f'training_histories/scenario_{i}_history_plot.png')
        plt.close()

plot_training_history(histories)

# === 15. Visualisasi Akurasi ===
plt.figure(figsize=(12, 6))
sns.barplot(x='scenario', y='accuracy', data=results_df, palette="mako")
plt.title("Akurasi Model per Skenario Eksperimen")
plt.xlabel("Nomor Skenario")
plt.ylabel("Akurasi")
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig("akurasi_per_skenario.png")
plt.show()
```
